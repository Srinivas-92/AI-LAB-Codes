```python
import heapq

def misplaced_tiles(board, goal):
    return sum(1 for i in range(9) if board[i] != 0 and board[i] != goal[i])

def manhattan_distance(board, goal):
    dist = 0
    for i in range(9):
        if board[i] != 0:
            r1, c1 = i // 3, i % 3
            g_idx = goal.index(board[i])
            r2, c2 = g_idx // 3, g_idx % 3
            dist += abs(r1 - r2) + abs(c1 - c2)
    return dist

def a_star_search(start, goal, heuristic_type):
    heuristic_fn = misplaced_tiles if heuristic_type == 1 else manhattan_distance
    queue = [(heuristic_fn(start, goal), 0, start, [])]
    visited = set()

    while queue:
        f, g, board, path = heapq.heappop(queue)

        if board == goal:
            return path + [board], g

        if board in visited:
            continue
        visited.add(board)

        blank = board.index(0)
        r, c = blank // 3, blank % 3
        for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
            nr, nc = r + dr, c + dc
            if 0 <= nr < 3 and 0 <= nc < 3:
                target = nr * 3 + nc
                temp = list(board)
                temp[blank], temp[target] = temp[target], temp[blank]
                next_board = tuple(temp)
                if next_board not in visited:
                    h = heuristic_fn(next_board, goal)
                    heapq.heappush(queue, (g + 1 + h, g + 1, next_board, path + [board]))
    return None, 0

def print_state(board):
    for i in range(0, 9, 3):
        print(" ".join(str(x) if x != 0 else "_" for x in board[i:i+3]))
    print()

start_state = (1, 2, 3, 5, 6, 0, 7, 8, 4)
goal_state = (1, 2, 3, 5, 8, 6, 0, 7, 4)

print("Initial State:")
print_state(start_state)
print("Goal State:")
print_state(goal_state)

print("Select Heuristic Method:")
print("1. Misplaced Tiles")
print("2. Manhattan Distance")
choice = int(input("Enter your choice (1 or 2): ") or "2")

solution, total_depth = a_star_search(start_state, goal_state, choice)
if solution:
    print(f"\nSolution found in {total_depth} moves!\n")
    for step_num, state in enumerate(solution):
        print(f"Step {step_num}:")
        print_state(state)
else:
    print("No solution exists.")
```
