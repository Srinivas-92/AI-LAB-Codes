```python
board = [' ' for _ in range(9)]

def print_board():
    for i in range(3):
        print(f'{board[i*3]}|{board[i*3+1]}|{board[i*3+2]}')
        if i < 2:
            print('-+-+-')

def check_win(player):
    win_conditions = [
        (0, 1, 2), (3, 4, 5), (6, 7, 8),
        (0, 3, 6), (1, 4, 7), (2, 5, 8),
        (0, 4, 8), (2, 4, 6)
    ]
    for cond in win_conditions:
        if all(board[i] == player for i in cond):
            return True
    return False

def check_draw():
    return ' ' not in board

def play_game():
    current_player = 'X'
    while True:
        print_board()
        try:
            position = int(input(f'Player {current_player}, choose a position (1-9): ')) - 1
            if not (0 <= position <= 8 and board[position] == ' '):
                print('Invalid move. Try again.')
                continue
        except ValueError:
            print('Invalid input. Please enter a number.')
            continue

        board[position] = current_player

        if check_win(current_player):
            print_board()
            print(f'Player {current_player} wins!')
            break
        if check_draw():
            print_board()
            print("It's a draw!")
            break

        current_player = 'O' if current_player == 'X' else 'X'

play_game()
```
