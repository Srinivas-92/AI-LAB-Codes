```python
facts = {
    "Barks(Tommy)",
    "FourLegs(Tommy)",
    "SharpTeeth(Tommy)"
}

def forward_chaining(facts):
    rules = True

    while rules:
        rules = False

        if "Barks(Tommy)" in facts and "FourLegs(Tommy)" in facts:
            if "Dog(Tommy)" not in facts:
                facts.add("Dog(Tommy)")
                rules = True

        if "Dog(Tommy)" in facts:
            if "Mammal(Tommy)" not in facts:
                facts.add("Mammal(Tommy)")
                rules = True

        if "Mammal(Tommy)" in facts and "SharpTeeth(Tommy)" in facts:
            if "Carnivore(Tommy)" not in facts:
                facts.add("Carnivore(Tommy)")
                rules = True

    return facts

facts = forward_chaining(facts)

print("Knowledge Base:")
for fact in sorted(facts):
    print(fact)

if "Carnivore(Tommy)" in facts:
    print("\nTommy is a Carnivore")
    print("Query Proved")
else:
    print("\nQuery Not Proved")
```
